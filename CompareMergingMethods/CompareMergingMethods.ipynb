# 3.) Compare old with new merging method for com1DFA

```python
# old/current merged dataframe
inputsDF_merged_com1
```

```python
# simHash is index, should not be index for merging
configurationDF_com1_new = configurationDF_com1.reset_index().rename(columns={'index': 'simHash'})

# necessary because simNAme appears twice, leads to renaming of both columns when merging
configurationDF_com1_new = configurationDF_com1_new.rename(columns={'simName': 'simName_renamed'})
#configurationDF__com1_new
```

```python
inputsDF_merged_com1_new = inputsDF_com1.reset_index().merge(configurationDF_com1_new, on=['simHash', 'modelType']).set_index('index')
inputsDF_merged_com1_new
```

```python
# drop redundant column
inputsDF_merged_com1_new = inputsDF_merged_com1_new.drop('simName_renamed', axis=1)
```

```python
# compare if 2 dF the same
inputsDF_merged_com1_new.sort_index(axis=1).equals(inputsDF_merged_com1.sort_index(axis=1))
```

## 3.1) Result:

The new merging method uses simHash as the key instead of simName. \
This method does not solve the unique simHash problem. \
For com1DFA, the output from the new and old merging methods remains identical. \
When merging two DataFrames on simHash, be aware that other columns — such as simName — may exist in both DataFrames. This can cause automatic renaming (e.g., simName_x, simName_y) during the merge. \
To avoid this:
- One approach is to manually rename the simName column in one of the DataFrames before merging.
- Alternatively, if the simName values are identical across both DataFrames, you can drop one of the redundant columns after merging.
