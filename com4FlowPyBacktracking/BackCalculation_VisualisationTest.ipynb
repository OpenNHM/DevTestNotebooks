#### define graph
- Here we define an example of a directed graph - a similar dir-graph structure can be obtained by com4FlowPy.
    - There is one start-node (this is the root of the graph - parent node of the whole graph)
    - from this start-node we have edges connecting the parent node(s) to their children
- We represent the graph by three dictionaries
    - topology - this dictionary stores the general graph topology - each key represents a node, while each item in the associated list-item represents an edge from the key-node (parent) to adjacent nodes (children)
    - values - here we store the values of infrastructure objects at each node
        - first this contains only values > 0 for nodes that are infrastructure objects
    - positions - positions of the nodes in (x,y) used for plotting


```python
#simple graph consisting of 14 nodes
topology = {0:  [1, 2, 3], # node 0 is the parent and has edges to nodes 1,2, and 3
            1:  [4],
            2:  [4, 5, 6],
            3:  [2, 6],
            4:  [7, 8],
            5:  [8, 9],
            6:  [10],
            7:  [11],        
            8:  [11],
            9:  [12],
            10: [13],
            11: [],        # node 11 is an end-node and does not have children
            12: [],
            13: [],}

values = {0: 0,
          1: 0,
          2: 1,
          3: 0,
          4: 2,
          5: 0,
          6: 1,
          7: 1,
          8: 0,
          9: 3,
          10:0,
          11:0,
          12:3,
          13:1}

positions = {0: (2,4),
             1: (1,3),
             2: (2,3),
             3: (3,3),
             4: (1,2),
             5: (2,2),
             6: (3,2),
             7: (0,1),
             8: (1,1),
             9: (2,1),
            10: (4,1),
            11: (0,0),
            12: (2,0),
            13: (4,0)}
```

```python
#plot initial graph topology
bC.plotGraph(topology, values, positions)
```

```python
#perform the backCalculation
#step 1 reverse graph
bC.plotGraph(bC.reverseTopology(topology), values, positions)
```

```python
# now we perform the backTracking and plot the results
valuesBackTrack = bC.backTracking(topology, values)
bC.plotGraph(bC.reverseTopology(topology), valuesBackTrack, positions)
```
